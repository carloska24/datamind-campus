## Pandas: filtrar e agrupar

**Objetivo:** Filtrar, ordenar e agrupar sem confundir seleção com mudança na tabela original.

Um filtro produz uma máscara booleana. Use & para combinar condições e coloque cada condição entre parênteses. sort_values organiza linhas. groupby reúne categorias para calcular medidas. mean é média; median é mediana; std mede dispersão. A média sozinha pode esconder diferenças entre produtos.

### Tente antes da solução

### Soluções para conferir
Execute uma célula por vez e interprete a saída.

```python
df[df['temperatura'] > 75]
```

```python
df[(df['temperatura'] > 75) & (df['defeitos'] > 10)]
```

```python
df.sort_values('temperatura', ascending=False)
```
