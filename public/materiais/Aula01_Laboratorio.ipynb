```python
import pandas as pd

df = pd.DataFrame({'produto': ['Motor A', 'Motor B', 'Motor A', 'Motor C', 'Motor B'], 'producao': [100, 120, 110, 90, 130], 'defeitos': [5, 12, 6, 3, 15], 'temperatura': [70, 82, 73, 68, 85]})
m = pd.DataFrame({'maquina': ['A', 'B', 'C', 'D', 'E'], 'producao': [1250, 1380, 1190, 1450, 1320], 'energia': [420, 465, 390, 510, 405]})
df
```

## KDD: transformação
Qual é a taxa de 12 em 120?

Transformar pode significar converter tipos, padronizar unidades, codificar categorias, agregar períodos ou criar indicadores. Normalização reescala valores e pode ajudar alguns algoritmos; não é uma obrigação para toda análise. Não confunda transformação com descoberta de padrões.

```python
df['taxa_defeitos'] = df['defeitos'] / df['producao'] * 100
```

```python
import pandas as pd
df.head()
df.shape
df.info()
df.describe()
df['temperatura']
df[['produto', 'temperatura']]
```

## Pandas: filtrar e agrupar
Qual a média de produção do Motor A?

Um filtro produz uma máscara booleana. Use & para combinar condições e coloque cada condição entre parênteses. sort_values organiza linhas. groupby reúne categorias para calcular medidas. mean é média; median é mediana; std mede dispersão. A média sozinha pode esconder diferenças entre produtos.

```python
df[df['temperatura'] > 75]
df[(df['temperatura'] > 75) & (df['defeitos'] > 10)]
df.sort_values('temperatura', ascending=False)
df.groupby('produto')['producao'].mean()
df.groupby('produto').agg(
    producao=('producao','mean'),
    defeitos=('defeitos','mean'))
```

## Laboratório 1: temperatura e defeitos
Qual frase é defensável: “temperatura causa defeitos” ou “nesta amostra variam juntas”?

Com os dados transcritos do slide: temperatura média 75,6; produção média 110; temperatura mínima 68 e máxima 85; defeitos médios 8,2. A correlação de Pearson entre temperatura e contagem de defeitos é 0.995051. Ela mede associação linear e não demonstra causalidade. Cinco registros são insuficientes para uma conclusão industrial robusta.

```python
df[['producao', 'defeitos', 'temperatura']].corr()
df[['temperatura', 'taxa_defeitos']].corr()
```

## Laboratório 2: produção e energia
Por que escolher D como mais eficiente seria um erro?

Dados por turno: A = 1250 peças / 420 kWh; B = 1380 / 465; C = 1190 / 390; D = 1450 / 510; E = 1320 / 405. A razão produção/energia tem unidade peças/kWh. Considere produto, período e condições equivalentes para comparar máquinas na operação real.

```python
m['eficiencia'] = m['producao'] / m['energia']
m.sort_values('eficiencia', ascending=False)
m['producao'].mean()
m['energia'].mean()
```

```python
print(m.loc[m['producao'].idxmax()])
print(m.loc[m['energia'].idxmin()])
print(m.loc[m['eficiencia'].idxmax()])
print(df.groupby('produto')['defeitos'].mean())
print(df.groupby('produto').agg({'producao':'mean','defeitos':'mean','temperatura':'mean'}))
```
